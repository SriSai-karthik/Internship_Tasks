
## Q1. Find the Frequency of Each Word in a Sentence

### Pseudocode

1. Take a sentence as input.
2. Split the sentence into individual words.
3. Create an empty dictionary to store word frequencies.
4. Read each word one by one.
5. If the word already exists in the dictionary, increase its count by 1.
6. Otherwise, add the word with a count of 1.
7. Return the dictionary.

```python
def word_frequency(sentence):
    words = sentence.split()
    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    return frequency


# Given test case
print(word_frequency("python is easy and python is powerful"))

# Additional test case
print(word_frequency("cat dog cat"))
```


## Q2. Find the Frequency of Each Character in a Word

### Pseudocode

1. Take a word as input.
2. Create an empty dictionary to store character frequencies.
3. Read each character in the word one by one.
4. If the character already exists in the dictionary, increase its count by 1.
5. Otherwise, add the character to the dictionary with a count of 1.
6. Return the dictionary.

```python
def character_frequency(word):
    frequency = {}

    for char in word:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    return frequency


# Given test case
print(character_frequency("banana"))

# Additional test case
print(character_frequency("hello"))
```


## Q3. Find the First Non-Repeating Character in a Word

### Pseudocode

1. Take a word as input.
2. Create an empty dictionary.
3. Count the frequency of each character in the word.
4. Read the original word again from left to right.
5. Check the frequency of each character.
6. If a character has a frequency of 1, return that character.
7. If no character has a frequency of 1, return None.

```python
def first_non_repeating(word):
    frequency = {}

    for char in word:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    for char in word:
        if frequency[char] == 1:
            return char

    return None


# Given test case
print(first_non_repeating("swiss"))

# Additional test case
print(first_non_repeating("aabb"))
```


## Q4. Check Whether Two Strings Are Anagrams

### Pseudocode

1. Take two strings as input.
2. If their lengths are different, return False.
3. Create an empty dictionary.
4. Count each character in the first string.
5. Read each character in the second string and decrease its count in the dictionary.
6. If a character is missing or any final count is not zero, return False.
7. Otherwise, return True.

```python
def are_anagrams(s1, s2):
    if len(s1) != len(s2):
        return False

    frequency = {}

    for char in s1:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    for char in s2:
        if char in frequency:
            frequency[char] -= 1
        else:
            return False

    for count in frequency.values():
        if count != 0:
            return False

    return True


# Given test case
print(are_anagrams("listen", "silent"))

# Additional test case
print(are_anagrams("hello", "world"))
```


## Q5. Find the First Repeating Character in a String

### Pseudocode

1. Take a string as input.
2. Create an empty set to store characters already seen.
3. Read each character from left to right.
4. If the character is already in the set, return that character.
5. Otherwise, add the character to the set.
6. If no character repeats, return None.

```python
def first_repeating_char(word):
    seen = set()

    for char in word:
        if char in seen:
            return char
        seen.add(char)

    return None


# Given test case
print(first_repeating_char("programming"))

# Additional test case
print(first_repeating_char("abcd"))
```


## Q6. Move All Zeros to the End of a List

### Pseudocode

1. Take a list of numbers as input.
2. Create an empty list to store non-zero numbers.
3. Read each number in the original list.
4. If the number is not zero, add it to the result list.
5. Calculate how many zeros were present in the original list.
6. Add those zeros to the end of the result list.
7. Return the result list.

```python
def move_zeros(nums):
    result = []

    for num in nums:
        if num != 0:
            result.append(num)

    zero_count = len(nums) - len(result)
    result.extend([0] * zero_count)

    return result


# Given test case
print(move_zeros([0, 3, 0, 1, 2, 0, 4]))

# Additional test case
print(move_zeros([0, 5, 0, 2]))
```


## Q7. Find the Second Highest Distinct Number in a List

### Pseudocode

1. Take a list of numbers as input.
2. Create a new list containing only distinct numbers.
3. If fewer than two distinct numbers exist, return None.
4. Sort the distinct numbers in descending order.
5. Return the number at the second position.

```python
def second_highest(nums):
    unique_nums = []

    for num in nums:
        if num not in unique_nums:
            unique_nums.append(num)

    if len(unique_nums) < 2:
        return None

    unique_nums.sort(reverse=True)

    return unique_nums[1]


# Given test case
print(second_highest([10, 20, 40, 30, 40, 20]))

# Additional test case
print(second_highest([5, 5, 3]))
```


## Q8. Remove Duplicates Without Changing the Original Order

### Pseudocode

1. Take a list as input.
2. Create an empty list to store unique elements.
3. Read each element in the original list.
4. If the element is not already in the result list, add it.
5. Return the result list.

```python
def remove_duplicates(nums):
    result = []

    for num in nums:
        if num not in result:
            result.append(num)

    return result


# Given test case
print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))

# Additional test case
print(remove_duplicates([1, 1, 2, 3, 2]))
```


## Q9. Find the Missing Number from 1 to N

### Pseudocode

1. Take a list of numbers and the value N as input.
2. Calculate the expected sum of numbers from 1 to N using the formula N × (N + 1) / 2.
3. Calculate the actual sum of the numbers in the list.
4. Subtract the actual sum from the expected sum.
5. Return the difference as the missing number.

```python
def find_missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)

    return expected_sum - actual_sum


# Given test case
print(find_missing_number([1, 2, 4, 5, 6], 6))

# Additional test case
print(find_missing_number([1, 3, 4, 5], 5))
```


## Q10. Find Two Numbers Whose Sum Equals a Target

### Pseudocode

1. Take a list of numbers and a target value as input.
2. Start with the first number in the list.
3. Compare it with each number that comes after it.
4. If the two numbers add up to the target, return them as a list.
5. Continue checking pairs in the required order until a valid pair is found.
6. If no valid pair exists, return an empty list.

```python
def find_pair(nums, target):
    for j in range(1, len(nums)):
        for i in range(j):
            if nums[i] + nums[j] == target:
                return [nums[i], nums[j]]
    return []


# Given test case
print(find_pair([2, 7, 11, 15], 9))

# Additional test case
print(find_pair([1, 4, 6, 8], 10))

# Test case 2 ---> to verify the earliest second element
print(find_pair([1, 4, 2, 5], 6))
```

## Q11. Find the Number Closest to Zero

**Pseudocode:**
1. Assume the first number is the closest to zero.
2. Check each number in the list.
3. If a number is closer to zero, update the closest number.
4. If two numbers are equally close to zero, choose the positive number.
5. Return the closest number.

```python
def closest_to_zero(nums):
    return min(nums, key=lambda num: (abs(num), -num))


# Given test case
print(closest_to_zero([-8, -3, 2, 5, -2]))

# Additional test case
print(closest_to_zero([-4, 4, -7]))
```

## Q12. Find Common Elements Between Two Lists

**Pseudocode:**
1. Create an empty result list.
2. Go through each element in the first list.
3. Check whether the element exists in the second list.
4. Check whether the element is not already in the result list.
5. If both conditions are true, add the element to the result list.
6. Return the result list.

```python
def common_elements(list1, list2):
    result = []
    for item in list1:
        if item in list2 and item not in result:
            result.append(item)
    return result


# Given test case
print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))

# Additional test case
print(common_elements([4, 2, 4, 1], [1, 4, 5]))
```

## Q13. Rotate a List to the Right by K Positions

**Pseudocode:**
1. Check whether the list is empty. If it is, return an empty list.
2. Calculate the effective number of rotations using the remainder of K divided by the list length.
3. If the effective rotation count is zero, return a copy of the original list.
4. Take the last K elements of the list.
5. Take the remaining elements from the beginning of the list.
6. Combine the two parts and return the rotated list.

```python
def rotate_right(nums, k):
    if not nums:
        return []

    k = k % len(nums)

    if k == 0:
        return nums[:]

    return nums[-k:] + nums[:-k]


# Given test case
print(rotate_right([1, 2, 3, 4, 5], 2))

# Additional test case
print(rotate_right([1, 2, 3], 1))
```

## Q14. Find the Maximum Sum of Two Consecutive Elements

**Pseudocode:**
1. Calculate the sum of the first two elements and store it as the maximum sum.
2. Go through each remaining pair of consecutive elements.
3. Calculate the sum of the current pair.
4. If the current sum is greater than the maximum sum, update the maximum sum.
5. Return the maximum sum.

```python
def max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]

    for i in range(1, len(nums) - 1):
        current_sum = nums[i] + nums[i + 1]

        if current_sum > max_sum:
            max_sum = current_sum

    return max_sum


# Given test case
print(max_consecutive_sum([4, 7, 2, 9, 5]))

# Additional test case
print(max_consecutive_sum([1, 5, 2]))
```

## Q15. Find All Pairs of Numbers Whose Sum Equals a Target

**Pseudocode:**
1. Create an empty list to store pairs.
2. Select each number in the list.
3. Compare it with every number that appears after it.
4. If the two numbers add up to the target, sort the pair in ascending order.
5. Add the pair only if it is not already in the result list.
6. Sort the result list by the first element of each pair.
7. Return the result list. If no valid pairs exist, return an empty list.

```python
def all_pairs_target(nums, target):
    pairs = []

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = sorted([nums[i], nums[j]])

                if pair not in pairs:
                    pairs.append(pair)

    pairs.sort()
    return pairs


# Given test case
print(all_pairs_target([2, 4, 3, 5, 7, 8, 1], 9))

# Additional test case
print(all_pairs_target([1, 2, 3, 4, 5], 6))
```
