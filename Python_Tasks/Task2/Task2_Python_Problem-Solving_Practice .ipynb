```python
# Q1. Find the Shortest Word in a Sentence
# Write a Python program to find the shortest word in a given sentence.
# Example input:
# Python makes coding easy
# Expected output:
# easy
```

### Observe
The input is a sentence, for example: "Python makes coding easy".
The expected output is the shortest word in that sentence. 
Looking at the words: Python (6 letters), makes (5 letters), coding (6 letters), easy (4 letters).
The shortest word is "easy".

### Understand
The problem is asking me to identify the word with the least number of characters.
I must not hardcode the answer. I should split the sentence into words and compare their lengths.
If two words have the same minimum length, I will return the first one that appears.

### Break
steps
1. Split the sentence into words using split()
2. Assume the first word is the shortest
3. Loop through each word
4. Compare its length with the current shortest.
5. If it is shorter, update the shortest word
6. Print the shortest word at the end

### Logic
I will use a loop to go through each word,then I will keep a variable called shortest, starting with the first word.

If I find a word shorter than shortest, I will replace it. Finally, I will print the shortest word.

```python
sentence = "Python makes coding easy"
words = sentence.split()
shortest = words[0]

for i in words:
    if len(i) < len(shortest):
        shortest = i
        
print(shortest)
```

### Test
- Input: "Python makes coding easy" → Output: easy
- Input: "I love data science" → Output: I
  
#### Edge case: If two words have the same length, the first one will be returned.

If it is a letter and not in vowels, increment consonant count.

```python
text = "Python 123!"
vowels = "aeiouAEIOU"
count = 0

for ch in text:
    if ch.isalpha():
        if ch not in vowels:
            count += 1


print(count)
```

```python
# Q8. Find the Most Frequent Character
# Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.
# Example input:
# banana
# Expected output:
# a
```

### Observe
The input is a string, for example: "banana".
The expected output is the character that appears most frequently.
In "banana", 'a' appears 3 times, 'n' appears 2 times, 'b' appears once.
So the output is 'a'.

### Understand
The problem is asking me to find the most frequent character.
If multiple characters have the same highest frequency, return the first one encountered.
Spaces should be ignored.

### Break
steps
1. Remove spaces from the string
2. Count frequency of each character using a dictionary
3. Loop through the string in order
4. Track the character with the highest frequency
5. Return the first one if there is a tie.

### Logic
I will build a frequency dictionary.
Then I will loop through the string again to preserve order.
The first character with the maximum frequency will be the answer.

```python
text = "banana"
freq = {}

for ch in text.replace(" ", ""):
    freq[ch] = freq.get(ch, 0) + 1


max_char = None
max_count = 0

for ch in text.replace(" ", ""):
    if freq[ch] > max_count:
        max_count = freq[ch]
        max_char = ch


print(max_char)
```

### Test:
- Input: "banana" → Output: a
- Input: "success" → Output: s

#### Edge case: If all characters appear once, the first character will be returned.

steps

3. Loop through the string in order

```python
text = "programming"
freq = {}

for ch in text.replace(" ", ""):
    freq[ch] = freq.get(ch, 0) + 1


result = []
seen = set()

for ch in text.replace(" ", ""):
    if freq[ch] > 1 and ch not in seen:
        result.append(ch)
        seen.add(ch)


print(result)
```

### Break
steps
1. Convert the sentence to lowercase
2. Split the sentence into words
3. Count the frequency of each word using a dictionary
4. Loop through the words in order to preserve the first occurrence
5. Track the word with the highest frequency

### Logic

### Logic
I will build a frequency dictionary, Then loop through the words again to preserve order.
If a word appears once and is longer than the current longest, update it

```python
sentence = "Python makes coding useful Python"
words = sentence.lower().split()
freq = {}

for w in words:
    freq[w] = freq.get(w, 0) + 1


longest = ""


for w in words:
    if freq[w] == 1 and len(w) > len(longest):
        longest = w

print(longest)
```

### Test
- Input: "Python makes coding useful Python" → Output: coding
- Input: "apple banana mango banana apple" → Output: mango

#### Edge case: If no word appears only once, output will be empty.
