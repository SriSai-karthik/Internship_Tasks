```python
# Q35

# import copy

# a = [[1, 2], [3, 4]]
# b = copy.copy(a)
# b[0][1] = 99
# print(a)
```

```python
# Prediction

# [[1, 99], [3, 4]]
```

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)

b[0][1] = 99

print(a)
```

```python
# Explanation

# copy.copy() makes a shallow copy. The outer list is new, but inner lists are shared.
# so, modifying b[0][1] changes a[0][1] too.
```

```python
# Q36

# import copy

# a = [[1, 2], [3, 4]]
# b = copy.deepcopy(a)
# b[0][1] = 99
# print(a)
```

```python
# Prediction

# [[1, 2], [3, 4]]
```

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)

b[0][1] = 99

print(a)
```

```python
# Expalantion

# deepcopy duplicates everything, inner lists are independent. so changing b doesn’t affect a
```
